```python

# ==========================================
# PACKETSCOPE - NETWORK ANALYSIS DASHBOARD
# ==========================================

# 1. INSTALL REQUIRED LIBRARIES AUTOMATICALLY

import subprocess
import sys
import importlib.util

packages = {
    "streamlit": "streamlit",
    "pandas": "pandas",
    "plotly": "plotly"
}

for module, package in packages.items():
    if importlib.util.find_spec(module) is None:
        print(f"Installing {package}...")
        subprocess.check_call([
            sys.executable, "-m", "pip", "install", package
        ])

# 2. IMPORT LIBRARIES

import streamlit as st
import pandas as pd
import plotly.express as px
from pathlib import Path

# 3. PAGE CONFIGURATION

st.set_page_config(
    page_title="PacketScope",
    page_icon="🌐",
    layout="wide"
)

st.title("🌐 PacketScope")
st.subheader("Network Traffic Analysis Dashboard")

st.write(
    "Explore network protocols, domains, destinations "
    "and traffic patterns captured from your device."
)

# 4. LOAD DATASETS

DATA_DIR = Path(__file__).parent / "data" / "dashboard"

@st.cache_data
def load_data(filename):
    path = DATA_DIR / filename

    if path.exists():
        return pd.read_csv(path)

    return pd.DataFrame()

tls_df = load_data("tls_domains.csv")
domain_df = load_data("domain_summary.csv")
dns_df = load_data("dns_summary.csv")
quic_df = load_data("quic_summary.csv")
destination_df = load_data("destination_summary.csv")
protocol_df = load_data("protocol_comparison.csv")
timeline_df = load_data("traffic_timeline.csv")
packet_size_df = load_data("packet_size_summary.csv")

# 5. SIDEBAR NAVIGATION

st.sidebar.title("PacketScope")

page = st.sidebar.radio(
    "Choose Analysis",
    [
        "Overview",
        "Domain Analysis",
        "Protocol Analysis",
        "Network Destinations",
        "Traffic Timeline",
        "Packet Size Analysis"
    ]
)

# 6. OVERVIEW

if page == "Overview":

    st.header("Network Overview")

    col1, col2, col3 = st.columns(3)

    total_tls = len(tls_df)
    total_dns = (
        dns_df["DNS_Requests"].sum()
        if "DNS_Requests" in dns_df.columns else 0
    )
    total_destinations = len(destination_df)

    col1.metric("TLS Observations", total_tls)
    col2.metric("DNS Requests", int(total_dns))
    col3.metric("Remote IP Addresses", total_destinations)

    st.divider()

    st.subheader("Protocol Distribution")

    if not protocol_df.empty:
        fig = px.pie(
            protocol_df,
            names="Protocol",
            values="Packets",
            hole=0.45
        )
        st.plotly_chart(fig, use_container_width=True)

# 7. DOMAIN ANALYSIS

elif page == "Domain Analysis":

    st.header("Domain Analysis")

    if not domain_df.empty:

        fig = px.bar(
            domain_df.head(15),
            x="Observations",
            y="Domain",
            orientation="h",
            title="Top Detected TLS Domains"
        )

        st.plotly_chart(fig, use_container_width=True)
        st.dataframe(domain_df, use_container_width=True)

    else:
        st.warning("Domain data unavailable.")

# 8. PROTOCOL ANALYSIS

elif page == "Protocol Analysis":

    st.header("Protocol Analysis")

    if not protocol_df.empty:

        fig = px.pie(
            protocol_df,
            names="Protocol",
            values="Packets",
            hole=0.4
        )

        st.plotly_chart(fig, use_container_width=True)
        st.dataframe(protocol_df, use_container_width=True)

# 9. NETWORK DESTINATIONS

elif page == "Network Destinations":

    st.header("Network Destinations")

    if not destination_df.empty:

        destination_df["Display"] = (
            destination_df["Domain"].fillna("Domain not detected")
            + " (" + destination_df["Remote IP"].astype(str) + ")"
        )

        fig = px.bar(
            destination_df.head(15).sort_values("Total_Packets"),
            x="Total_Packets",
            y="Display",
            orientation="h",
            title="Top 15 Network Destinations"
        )

        st.plotly_chart(fig, use_container_width=True)
        st.dataframe(destination_df, use_container_width=True)

# 10. TRAFFIC TIMELINE

elif page == "Traffic Timeline":

    st.header("Network Traffic Timeline")

    if not timeline_df.empty:

        timeline_df["Second"] = pd.to_datetime(
            timeline_df["Second"]
        )

        fig = px.line(
            timeline_df,
            x="Second",
            y="Packets",
            color="Direction",
            title="Incoming vs Outgoing Traffic"
        )

        st.plotly_chart(fig, use_container_width=True)

# 11. PACKET SIZE ANALYSIS

elif page == "Packet Size Analysis":

    st.header("Packet Size Analysis")

    if not packet_size_df.empty:

        fig = px.bar(
            packet_size_df,
            x="Average_Size",
            y="Domain",
            orientation="h",
            title="Average Packet Size by Domain"
        )

        st.plotly_chart(fig, use_container_width=True)
        st.dataframe(packet_size_df, use_container_width=True)

    else:
        st.warning("Packet-size data unavailable.")
```
